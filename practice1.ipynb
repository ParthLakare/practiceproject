```python
age_median = df['age'].median()
df['age'] = df['age'].fillna(age_median)
```

```python
df.drop(columns=['therapy_history'], inplace=True)
```

```python
df.groupby('age_started_drinking')['employment_status'].count()
```

```python
df['education_level']=df['education_level'].fillna(0)
```

```python
df['education_level'].unique()
```

```python
plt.figure(figsize=(10,7))
plt.bar(df['country'], df['employment_status'],color='skyblue')
plt.xlabel('Country')
plt.ylabel('Employment Status')  
plt.title('Employment Status by Country')
plt.xticks(rotation=40)
plt.legend(['Employment Status'])
plt.show()
```

```python
header
```

```python
plt.figure(figsize=(10,7))
plt.bar(df['country'], df['exercise_frequency'], color='orange', width=0.5)
plt.xlabel('Country')
plt.ylabel('Exercise Frequency')
plt.title('Exercise Frequency by Country')
plt.xticks(rotation=70)
plt.legend(['Exercise Frequency'], loc='upper left')
plt.show()
```

```python
plt.pie(df['education_level'].value_counts(), 
        labels=df['education_level'].unique(), 
        autopct='%1.1f%%', 
        startangle=90,
        shadow=True,
        )
plt.title('Education Level Distribution')
plt.axis('equal') 
```

```python
plt.pie(df['mental_health_status'].value_counts(), 
        labels=df['mental_health_status'].unique(), 
        autopct='%1.1f%%', 
        startangle=90, 
        shadow=True,
        colors=["#ff999936","#0dd89f","#cc99ff","#cc1224"],)
```
